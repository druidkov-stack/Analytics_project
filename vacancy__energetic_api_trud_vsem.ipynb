# 16. Второй вариант (Парсинг дает только 10000 вакансий, потом идет защита и сервер выдает код 500) Пришлось прибегнуть к помощи ИИ, а также ввести паузу и разбить парсинг на 2 этапа

```python
region_code = 74

dct_one = {
    'job-name': None,
    'salary': None,
    'salary_min': None,
    'salary_max': None,
    'company': None,
    'addresses': None,
    'vac_url': None
}

def clean_text(text):
    """Очистка текста от проблемных символов"""
    if text is None:
        return ''
    if not isinstance(text, str):
        text = str(text)
    return text.strip()

def parse_vacancies(start_offset, end_offset, limit=100):
    """Парсинг вакансий с заданного диапазона offset"""
    lst_dict = []
    
    for offset_number in tqdm(range(start_offset, end_offset), desc=f"Парсинг offset {start_offset}-{end_offset-1}"):
        link_pag = f'http://opendata.trudvsem.ru/api/v1/vacancies/region/74?offset={offset_number}'
        
        try:
            answer = requests.get(link_pag, timeout=30)
            time.sleep(3)  # Задержка между запросами
            
            if answer.status_code == 200:
                dct_vacs = answer.json()
                vacancy_data = dct_vacs.get('results', {}).get('vacancies', [])
                
                for vacancy_item in vacancy_data:
                    vacancy_dict = dct_one.copy()
                    vac = vacancy_item.get('vacancy', {})
                    
                    # Простые поля
                    vacancy_dict['job-name'] = clean_text(vac.get('job-name'))
                    vacancy_dict['salary'] = clean_text(vac.get('salary'))
                    vacancy_dict['salary_min'] = clean_text(vac.get('salary_min'))
                    vacancy_dict['salary_max'] = clean_text(vac.get('salary_max'))
                    vacancy_dict['vac_url'] = clean_text(vac.get('vac_url'))
                    
                    # Извлекаем название компании
                    company = vac.get('company')
                    vacancy_dict['company_name'] = clean_text(company.get('name')) if company else ''
                    
                    # Извлекаем адрес
                    addresses = vac.get('addresses')
                    if addresses and addresses.get('address'):
                        vacancy_dict['address_location'] = clean_text(addresses['address'][0].get('location'))
                    else:
                        vacancy_dict['address_location'] = ''
                    
                    lst_dict.append(vacancy_dict)
            else:
                print(f"Ошибка {answer.status_code} на offset {offset_number}")
                
        except Exception as e:
            print(f"Ошибка при загрузке offset {offset_number}: {e}")
            continue
    
    return lst_dict

def save_to_csv(data, filename, first_write=True):
    """Сохранение данных в CSV"""
    if not data:
        print(f"Нет данных для сохранения в {filename}")
        return
    
    fieldnames = list(data[0].keys())
    mode = 'w' if first_write else 'a'
    
    with open(filename, mode, newline='', encoding='utf-8-sig') as csvfile:
        writer = csv.DictWriter(csvfile, fieldnames=fieldnames, delimiter=';', quotechar='"', quoting=csv.QUOTE_ALL)
        if first_write:
            writer.writeheader()
        writer.writerows(data)
    
    print(f"Сохранено {len(data)} записей в {filename}")

# ============================================
# ГЛАВНАЯ ПРОГРАММА - ПАРСИНГ В ДВА ЭТАПА
# ============================================

# Имена файлов
file_part1 = 'вакансии_74_регион_part1.csv'
file_part2 = 'вакансии_74_регион_part2.csv'
file_total = 'вакансии_74_регион_total.csv'

# Удаляем старые файлы, если существуют
for file in [file_part1, file_part2, file_total]:
    if os.path.isfile(file):
        os.remove(file)
        print(f"Старый файл {file} удален")

print("=== НАЧАЛО ПАРСИНГА ===")

# ЭТАП 1: Парсинг offset с 0 по 60 (первые 61 страница)
print("\n--- ЭТАП 1: offset 0-60 ---")
data_part1 = parse_vacancies(0, 61)
save_to_csv(data_part1, file_part1, first_write=True)

# Пауза между этапами для восстановления сервера
print("\nПауза 300 секунд перед вторым этапом...")
time.sleep(300)

# ЭТАП 2: Парсинг offset с 61 по 122 (оставшиеся 62 страницы)
print("\n--- ЭТАП 2: offset 61-122 ---")
data_part2 = parse_vacancies(61, 123)
save_to_csv(data_part2, file_part2, first_write=True)

# ============================================
# ОБЪЕДИНЕНИЕ ДАННЫХ В ИТОГОВЫЙ ФАЙЛ
# ============================================
print("\n--- ОБЪЕДИНЕНИЕ ДАННЫХ ---")
combined_data = data_part1 + data_part2
save_to_csv(combined_data, file_total, first_write=True)

# ============================================
# ИТОГИ
# ============================================
print("\n=== ПАРСИНГ ЗАВЕРШЕН ===")
print(f"Часть 1: {len(data_part1)} записей -> {file_part1}")
print(f"Часть 2: {len(data_part2)} записей -> {file_part2}")
print(f"Всего: {len(data_part1) + len(data_part2)} записей -> {file_total}")
print(f"\nСозданы файлы:")
print(f"  - {file_part1}")
print(f"  - {file_part2}")
print(f"  - {file_total} (объединенный)")

# Проверяем содержимое файлов
for file in [file_part1, file_part2, file_total]:
    if os.path.isfile(file):
        with open(file, 'r', encoding='utf-8-sig') as f:
            lines = f.readlines()
            print(f"\n{file}: {len(lines) - 1} записей (всего строк: {len(lines)})")
```
